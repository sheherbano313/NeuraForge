Definition

The mean is the arithmetic average.

Formula
$$ Mean = \frac{\sum x}{n} $$

```python
data = [10, 20, 30, 40, 50]

mean = np.mean(data)

print("Mean:", mean)
```

```python
#The median is the middle value after sorting the data.

data = [10, 20, 30, 40, 50]

median = np.median(data)

print("Median:", median)
```

```python
#For an even number:

data = [10, 20, 30, 40]

print(np.median(data))
```

```python
#The mode is the most frequently occurring value.
#import statistics as stats
from scipy import stats


data = [10, 20, 20, 30, 30, 30, 40]

mode = stats.mode(data, keepdims=True)

print("Mode:", mode.mode[0])
```

Range is the difference between the maximum and minimum values.

```python

data = [10, 20, 30, 40, 50]

data_range = np.max(data) - np.min(data)

print("Range:", data_range)
```

```python
data = [10, 20, 30, 40, 50]

print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

A weighted mean gives different importance/weight to different values.


```python

scores = [80, 70, 90]
weights = [0.20, 0.30, 0.50]

weighted_mean = np.average(
    scores,
    weights=weights
)

print("Weighted Mean:", weighted_mean)
```

10. Geometric Mean

The geometric mean is the nth root of the product of n values.

It is useful for growth rates and ratios.



```python
data = [2, 8]

geometric_mean = stats.gmean(data)

print("Geometric Mean:", geometric_mean)
```

A percentile tells us the value below which a certain percentage of observations falls.

```python


data = [10, 20, 30, 40, 50, 60, 70, 80, 90]

p25 = np.percentile(data, 25)
p50 = np.percentile(data, 50)
p75 = np.percentile(data, 75)

print("25th percentile:", p25)
print("50th percentile:", p50)
print("75th percentile:", p75)
```

Quartiles


Quartiles divide ordered data into four parts.



```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

Variance

Variance measures how far values are spread from the mean.

Population variance:

```python


data = [10, 20, 30, 40, 50]

population_variance = np.var(data)

print(population_variance)
```

```python
sample_variance = np.var(data, ddof=1)

print(sample_variance)
```

Standard Deviation 
Standard deviation measures the typical spread of values around the mean.

It is the square root of variance.

```python

data = [10, 20, 30, 40, 50]

population_std = np.std(data)

sample_std = np.std(data, ddof=1)

print("Population SD:", population_std)
print("Sample SD:", sample_std)
```

Z-Score 

A z-score tells us how many standard deviations a value is away from the mean.

Formula:

$$ z = \frac{x-\mu}{\sigma} $$


```python



data = [10, 20, 30, 40, 50]

z_scores = stats.zscore(data)

print(z_scores)
```

```python
#An outlier is an observation unusually far from the rest of the data.

data = [10, 12, 13, 15, 16, 18, 20, 100]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower:", lower_bound)
print("Upper:", upper_bound)
```

```python
outliers = [
    x for x in data
    if x < lower_bound or x > upper_bound
]

print("Outliers:", outliers)
```
