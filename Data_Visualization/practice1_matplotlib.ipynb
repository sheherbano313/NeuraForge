```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [10, 15, 12, 20, 25]
```

Practice 1 — Line Plot

```python
plt.plot(days,sales)

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Daily Sales")

plt.show()
```

Practice 2 — Bar Chart

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "Ahmed", "Ayesha"]
marks = [65, 80, 75, 90]

plt.bar(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

Practice 3 — Scatter Plot

```python
import matplotlib.pyplot as plt

hours = [2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 60, 65, 75, 85]

plt.scatter(hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

Practice 4 — Histogram

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```
