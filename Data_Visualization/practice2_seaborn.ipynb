```python
import matplotlib.pyplot as plt
import seaborn as sns

hours = [2, 3, 4, 5, 6]
marks = [40, 45, 55, 65, 80]

# Matplotlib
plt.plot(hours, marks)
plt.show()

# Seaborn
sns.scatterplot(x=hours, y=marks)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

departments = ["CS", "IT", "CS", "AI", "IT", "CS"]

sns.countplot(x=departments)
plt.show()
```

Practice 2: sns.histplot()

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90]

sns.histplot(x=marks, bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

Seaborn boxplot()

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 150]

sns.boxplot(x=marks)

plt.title("Marks")
plt.show()
```
